```python
import numpy as np
```

```python
#Q1
l = [int(x) for x in input("Enter the list of number :").split(",")]
mean_without = sum(l)/len(l)
print(f"The mean is {mean_without}")
```

```python
#Q2
l = [int(x) for x in input("Enter the list of number :").split(",")]
mean_with = np.mean(l)
print(f"The mean is {mean_with}")
```

```python
#Q3
l = [int(x) for x in input("Enter the list of number :").split(",")]
l.sort()
if len(l)%2 == 0:
    median_without = (l[len(l)//2] + l[len(l)//2 - 1])/2
else:
    median_without = l[len(l)//2]
print(f"The median is {median_without}")
```

```python
#Q4
l = [int(x) for x in input("Enter the list of number :").split(",")]
median_with = np.median(l)
print(f"The median is {median_with}")
```

```python
#Q5
l = [int(x) for x in input("Enter the list of number :").split(",")]
dict = {}
for x in l:
    if x in dict:
        dict[x] += 1
    else:
        dict[x] = 1
mode_without = max(dict, key=dict.get)
print(mode_without)
```

```python
#Q6
l = [int(x) for x in input("Enter the list of number :").split(",")]
mode_with = np.bincount(l).argmax()
print(mode_with)
```

```python
#Q7
l = [int(x) for x in input("Enter the list of number :").split(",")]
mean = sum(l)/len(l)
sum_squared_diff = sum((x - mean) ** 2 for x in l)
variance = sum_squared_diff / len(l)
print(f"Varience without using function : {variance:.4f}")
```

```python
#Q8
l = [int(x) for x in input("Enter the list of number :").split(",")]
l = np.array(l)
variance = np.var(l)
print(f"Varience with using function : {variance:.4f}")
```

```python
#Q9
l = [int(x) for x in input("Enter the list of number :").split(",")]
mean = sum(l)/len(l)
sum_squared_diff = sum((x - mean) ** 2 for x in l)
sd = (sum_squared_diff / len(l))**0.5
print(f"Standard Deviation without function : {sd:.4f}")

l2 = np.array(l)
sd = np.std(l2)
print(f"Standard Deviation with using function : {sd:.4f}")
```

```python
#Q10
l = [int(x) for x in input("Enter the list of number :").split(",")]
mean = sum(l)/len(l)
sum_squared_diff = sum((x - mean) ** 2 for x in l)
variance = sum_squared_diff / len(l)
print(f"Varience without using function : {variance:.4f}")
sd = variance**0.5
print(f"Standard Deviation without function : {sd:.4f}")
```
