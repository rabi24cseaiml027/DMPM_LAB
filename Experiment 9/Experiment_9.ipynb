```python
import pandas as pd

def reco(data, min_support):
    tran = [set(t) for t in data.values()]
    N = len(tran)
    min_count = min_support * N
    unique_items = sorted(list(set(item for t in tran for item in t)))
    curr_cand = [frozenset([item]) for item in unique_items]
    records = []
    while curr_cand:
        freq_k = []
        for cand in curr_cand:
            count = sum(1 for t in tran if cand.issubset(t))
            if count >= min_count:
                support = count / N
                freq_k.append(cand)
                records.append({
                    "Itemset Size": len(cand),
                    "Support": support,
                    "Frequent Itemset": f"{{{', '.join(sorted(list(cand)))}}}"
                })
        if not freq_k:
            break
        next_cand = set()
        for i in range(len(freq_k)):
            for j in range(i + 1, len(freq_k)):
                itemset1 = sorted(list(freq_k[i]))
                itemset2 = sorted(list(freq_k[j]))
                if itemset1[:-1] == itemset2[:-1]:
                    next_cand.add(freq_k[i] | freq_k[j])
        curr_cand = list(next_cand)
    df = pd.DataFrame(records)
    df["Support"] = (df["Support"] * 100).astype(int).astype(str) + "%"
    df = df.sort_values(by=["Itemset Size", "Frequent Itemset"]).reset_index(drop=True)
    return df
dataset = {
    "T1" : ["Bread", "Milk"],
    "T2" : ["Bread", "Diaper", "Beer", "Eggs"],
    "T3": ["Milk", "Diaper", "Beer", "Coke"],
    "T4": ["Bread", "Milk", "Diaper", "Beer"],
    "T5": ["Bread", "Milk", "Diaper", "Coke"]
}
min_support=0.6
res = reco(dataset, min_support)
print(res.to_string(index=False))
```

```python
import pandas as pd

idx = ["Strawberry", "Coffee", "Orange", "Vanilla"]
data = {"Male":[23,18,8,8], "Female":[15,6,12,10]}
df = pd.DataFrame(data, index=idx)

O = df.values.tolist()
R = len(O)
C = len(O[0])

r_tot = [sum(r) for r in O]
c_tot = [sum(O[i][j] for i in range(R)) for j in range(C)]
G = sum(r_tot)

E_mat = []
X2 = 0

for r in range(R):
    e_row = []
    for c in range(C):
        o_val = O[r][c]
        e_val = (r_tot[r] * c_tot[c]) / G
        e_row.append(e_val)
        X2 += ((o_val - e_val) ** 2) / e_val
    E_mat.append(e_row)

df_E = pd.DataFrame(E_mat, index=idx, columns=data.keys())
df = (R - 1) * (C - 1)

print("\n--- a. Hypotheses ---")
print("Null Hypothesis (H0): Chocolate cream flavor preference is INDEPENDENT of gender.")
print("Alternative Hypothesis (H1): Chocolate cream flavor preference is DEPENDENT on gender.\n")

print("--- b. Degrees of Freedom ---")
print(f"Degrees of freedom (df): {df}\n")

print("--- c. Expected Frequency Table ---")
print(df_E.round(2), "\n")

print("--- Full Statistical Summary ---")
print(f"Calculated Chi-Square Value (X²): {X2:.4f}")

tab_val = 7.815
print(f"Tabulated Chi-Square Value (α=0.05) : {tab_val:.4f}")

if X2 < tab_val:
    print("\nConclusion: Accept H0. Flavor preference is independent of gender.")
else:
    print("\nConclusion: Reject H0. Flavor preference is dependently associated with gender.")
```
