```python
#Q1
import numpy as np
l = np.array([10,20,30,40,50])

#z-score normalization
mean = l.mean()
sd= l.std()
print("The z-score normalization for : ")
for i in l:
  print(f"{i} is {(i-mean)/sd:.4f}")

#min-max normalization
min = l.min()
max = l.max()
print("\nThe min-max normalization for : ")
for i in l:
  print(f"{i} is {(i-min)/(max-min):.4f}")

#Decimal Scalling Normalization
j = len(str(max))
print("\nThe decimal scalling normalization for : ")
for i in l:
  print(f"{i} is {i/(10**j):.4f}")
```

```python
#Q4
import numpy as np
import pandas as pd
df = pd.read_csv("https://raw.githubusercontent.com/RajeshRanaGiet/DMPM-Lab/main/Experiment%205/Rain%20Prediction%20Train.csv")
data = np.array(df['WindGustSpeed'][:7].dropna().values)
mean = np.mean(data)
sd = np.std(data)
var = np.var(data)
print(f"The mean is {mean:.4f}")
print(f"The standard deviation is {sd:.4f}")
print(f"The variance is {var:.4f}")
print("\nThe z-score normalization for : ")
for i in data:
  print(f"{i} is {(i-mean)/sd:.4f}")
```
