```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns


# Load the dataset
df = pd.read_csv("https://raw.githubusercontent.com/RajeshRanaGiet/DMPM-Lab/main/Experiment%204/cancer%20dataset.csv")


# =========================================================
# 1. Box Plot: Compare Radius distribution by Diagnosis
# =========================================================
sns.boxplot(data=df, x="Diagnosis", y="Radius (mean)")
plt.title("1. Box Plot: Radius (mean) by Diagnosis")
plt.xlabel("Diagnosis (B = Benign, M = Malignant)")
plt.ylabel("Radius (mean)")
plt.show()


# =========================================================
# 2. Scatter Plot: Relationship between Radius & Texture
# =========================================================
sns.scatterplot(data=df, x="Radius (mean)", y="Texture (mean)",)
plt.title("2. Scatter Plot: Radius vs. Texture")
plt.xlabel("Radius (mean)")
plt.ylabel("Texture (mean)")
plt.show()


# =========================================================
# 3. Histogram: Distribution of Area (mean)
# =========================================================
sns.histplot(df["Area (mean)"])
plt.title("3. Histogram: Distribution of Area (mean)")
plt.xlabel("Area (mean)")
plt.ylabel("Count")
plt.show()


# =========================================================
# 4. Line Chart: Trend of Radius (mean) across samples
# =========================================================
plt.plot(df["Radius (mean)"].head(30), label="Radius (mean)",)
plt.title("4. Line Chart: Radius (mean) for First 30 Samples")
plt.xlabel("Sample Index")
plt.ylabel("Radius (mean)")
plt.grid(True)
plt.show()


# =========================================================
# 5. Pie Chart: Percentage breakdown of Diagnosis (B vs M)
# =========================================================
diagnosis_counts = df["Diagnosis"].value_counts()
plt.pie(diagnosis_counts, labels=diagnosis_counts.index, autopct="%1.1f%%")
plt.title("5. Pie Chart: Diagnosis Distribution")
plt.show()


# =========================================================
# 6. Bar Plot: Average Area (mean) by Diagnosis
# =========================================================
mean_area = df.groupby("Diagnosis")["Area (mean)"].mean().reset_index()
sns.barplot(data=mean_area, x="Diagnosis", y="Area (mean)")
plt.title("6. Bar Plot: Average Area by Diagnosis")
plt.xlabel("Diagnosis")
plt.ylabel("Average Area (mean)")
plt.show()
```
